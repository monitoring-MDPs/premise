```python
from argparse import Namespace

from stormpy.pycarl import Interval

from premise.interval.interval import create_monitor
from premise.interval.loading import load_imc

model_path = "../out/models/2025-06-05_10-54-53/airportA-7-10-10-comp-ref5"
target = "crash"

args = Namespace()
args.trans_path = model_path+"-interval.npy"
args.init_path = model_path+"-initial_interval.npy"
trans, init = load_imc(args)
mon, obs_map, _, pomdp = create_monitor(trans, init, "min", target, 15)
```

```python
from stormvogel import stormpy_to_stormvogel, show, layout

sv_pomdp = stormpy_to_stormvogel(pomdp)
for transitions in sv_pomdp.transitions.values():
    for t in transitions.transition.values():
        for i in range(len(t.branch)):
            interval: Interval = t.branch[i][0]
            s = t.branch[i][1]
            t.branch[i] = (float(interval.diameter()), s)
my_layout = layout.DEFAULT()
my_layout.layout["numbers"]["Fractions"] = False
show(sv_pomdp, show_editor=True, max_states=100000, layout=my_layout)
```
